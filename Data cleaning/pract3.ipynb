```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns
from sklearn.impute import SimpleImputer
```

```python
deli = pd.read_csv('deliveries.csv')
match = pd.read_csv('matches.csv')
```

```python
num_col_devli = deli.select_dtypes(include=['O']).columns
num_col_match = match.select_dtypes(include=['O']).columns
```

```python
impute_mean = SimpleImputer(strategy='most_frequent')
```

```python
impute_mean
```

```python
impute_mean.fit(deli[num_col_devli])
```

```python
impute_mean.statistics_
```

```python
deli[num_col_devli] = impute_mean.transform(deli[num_col_devli])
```

```python
impute_mean = SimpleImputer(strategy='most_frequent')
```

```python
impute_mean.fit(match[num_col_match])
```

```python
match[num_col_match] = impute_mean.transform(match[num_col_match])
```
