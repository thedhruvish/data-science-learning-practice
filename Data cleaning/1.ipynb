```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.impute import SimpleImputer
```

```python
df= pd.read_csv("DATASET1.csv")
```

```python
x_train = df.drop(columns='SalePrice')
y_train = df['SalePrice']
```

```python
num_col = x_train.select_dtypes(include=['int64','float64']).columns
num_col
```

```python
x_train[num_col].isnull().sum()
```

```python
impute_mean=SimpleImputer(strategy='mean')
```

```python
impute_mean
```

```python
impute_mean.fit(x_train[num_col])
```

```python
impute_mean.statistics_
```

```python
impute_mean.transform(x_train[num_col])
```

```python
x_train[num_col].isnull().sum()
```

```python
impute_mean=SimpleImputer(strategy='continue',fill_value=90)
```

```python
impute_mean
```

```python
cat_col = x_train.select_dtypes(include="O").columns
```

```python
cat_col
```

```python
x_train[cat_col].isnull().sum()
```

```python
impute_mean=SimpleImputer(strategy='most_frequent')
```
