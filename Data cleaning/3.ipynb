```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 
from sklearn.impute import SimpleImputer 
from sklearn.compose import ColumnTransformer 
from sklearn.pipeline import Pipeline
```

```python
train=pd.read_csv('DATASET1.csv')
test=pd.read_csv('DATASET2.csv')
```

```python
x_train=train.drop(columns=['SalePrice'],axis=1)
```

```python
null_var=x_train.isnull().sum()
```

```python
cat_col=x_train.select_dtypes(include="O").columns
```

```python
cat_col
```

```python
x_train.columns
```

```python
num_col_mean=['LotFrontage']
num_col_median =[ 'MasVnrArea', 'GarageYrBlt']
num_col_mode=['Alley','MasVnrType','BsmtQual','BsmtCond','BsmtExposure','BsmtFinType1', 'BsmtFinType2','Electrical', 'FireplaceQu'] 
cat_col_missing=['GarageType', 'GarageFinish', 'GarageQual', 'GarageCond', 'PoolQC', 'Fence', 'MiscFeature']
```

```python
num_col_mean_imputer= Pipeline(steps=[("imputer",SimpleImputer(strategy="mean"))])
num_col_median_imputer=Pipeline(steps=[("imputer",SimpleImputer(strategy="median"))])
cat_col_mode_imputer=Pipeline(steps=[("imputer",SimpleImputer(strategy="most_frequent"))])
cat_col_missing_imputer=Pipeline(steps=[("imputer",SimpleImputer(strategy="constant",fill_value="missing"))])
```

```python
preprocessor=ColumnTransformer(transformers=
[('mean_imputer',num_col_mean_imputer,num_col_mean),
('median_imputer',num_col_median_imputer,num_col_median),
('mode_imputer',cat_col_mode_imputer,num_col_mode),
('missing_imputer',cat_col_missing_imputer,cat_col_missing)])
```

```python
preprocessor.fit(x_train)
```

```python
preprocessor.transform
```

```python
preprocessor.named_transformers_["mean_imputer"].named_steps['imputer'].statistics_
```

```python
preprocessor.named_transformers_["mean_imputer"].named_steps['imputer'].statistics_
```

```python
x_train_clean=preprocessor.transform(x_train)
x_train_clean
```
