```python
q1 = df['residual sugar'].quantile(0.25)
q3 = df['residual sugar'].quantile(0.75)
iqr = q1 - q3
q1,q3,iqr
```

```python
upper_limtit = q3 + (1.5 * iqr)
lower_limtit = q1 - (1.5 * iqr)
upper_limtit,lower_limtit
```

```python
df.loc[(df['residual sugar'] > upper_limtit) | (df['residual sugar'] < lower_limtit)]
```

```python
new_df = df.loc[(df['residual sugar'] <= upper_limtit) &(df['residual sugar'] >= lower_limtit)]

print("before remove = ",len(df))
print("after remove = ",len(new_df))
print("outliers: ",len(df)-len(new_df))
```

```python
new_df.reset_index(inplace=True)
```

```python
sns.boxplot(new_df['residual sugar'])
```

```python
new_df = df.copy()
new_df.loc[(new_df['residual sugar'] > upper_limtit),'residual sugar'] = upper_limtit
new_df.loc[(new_df['residual sugar'] > lower_limtit),'residual sugar'] = lower_limtit
```

```python
sns.boxplot(new_df['residual sugar'])
```
