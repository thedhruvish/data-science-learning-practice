```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns
from sklearn.impute import SimpleImputer
```

```python
df = pd.read_csv("https://raw.githubusercontent.com/12345k/IPL-Dataset/refs/heads/master/IPL/data.csv")
```

```python
num_col = df.select_dtypes(include=['O']).columns
```

```python
num_col
```

```python
impute_mean =SimpleImputer(strategy='most_frequent')
```

```python
impute_mean
```

```python
impute_mean.fit(df[num_col])
```

```python
impute_mean.statistics_
```

```python
df[num_col] = impute_mean.transform(df[num_col])
```
